```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.size'] = 14
```

```python
k = 4.359886822599401
lamb = 236.76593258114204
coef_reg = 0

def hazard(x, k = 1.359886822599401, lamb = 236.76593258114204):
  return (k/lamb) * ((x/lamb) ** (k-1))

# CDF
plt.subplot(1, 2, 1)
max_data = 300
x = np.linspace(0, int(max_data), int(max_data) * 100 + 1)
color = ['red', 'orange', 'yellow', 'green', 'blue', 'cyan', 'magenta', 'purple', 'black']
for i in range(0, 9):
  plt.plot(x, [hazard(x, k=5+i) for x in x], color=color[i], label=f'k={i}')
plt.axvline(x=lamb, color='black', linestyle='--')
plt.xlabel("time")
plt.ylim(0, 0.04)
plt.title("hazard function")
plt.legend()
```

```python
def linear(x, rul):
  return lamb * (1 - (x/rul))

def exponential(x, rul):
  return lamb ** (1 - (x/rul)) - 1

def rev_exp(x, rul):
  return lamb - lamb ** ((x/rul))

# showing each function
max_data = 350
x = np.linspace(0, int(max_data), int(max_data) * 100 + 1)
plt.plot(x, [linear(x, rul=max_data) for x in x], color='red', label='Linear Degration')
# plt.plot(x, [exponential(x, rul=max_data) for x in x], color='blue', label='exponential')
# plt.plot(x, [rev_exp(x, rul=max_data) for x in x], color='yellow', label='reverse exponential')
plt.plot(x, max(x) - x, color='green', label='Actual RUL')
plt.axhline(y=0, color='black', linestyle='--')

plt.legend()
```
